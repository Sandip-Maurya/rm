```python
import torch
from qwen_tts import Qwen3TTSModel

print("GPU:", torch.cuda.get_device_name(0))
print("CUDA:", torch.version.cuda)

model = Qwen3TTSModel.from_pretrained(
    "Qwen/Qwen3-TTS-12Hz-1.7B-CustomVoice",
    device_map="cuda:0",
    dtype=torch.bfloat16,
)

print("✅ Qwen3-TTS loaded!")
```

```python
text = """
Python three point fifteen just dropped, and here are the top three game-changing features you need to know!

Number one: Lazy Imports.

Massive libraries like PyTorch or Pandas won't freeze your startup. Modules load only when they're first called.

Startup dropped from one point two seconds to forty milliseconds!

Number two: frozendict.

Python now has native immutable dictionaries!

They're thread-safe, lockable, and can finally be used directly as dictionary keys.

Number three: unpacking inside comprehensions!

You can now use star-unpacking directly inside list, set, and dictionary comprehensions.

No more messy itertools hacks.

Which one is your favorite? Let me know below, and follow for Part two, with UTF-8, faster JIT, and Tachyon!
"""

wavs, sr = model.generate_custom_voice(
    text=text,
    language="English",
    speaker="Ryan",
    instruct="High-energy, confident tech creator delivering a YouTube Shorts voiceover. Fast but very clear pace. Crisp pronunciation, punchy emphasis on feature names and numbers, energetic transitions, natural conversational delivery. Avoid sounding robotic or overly dramatic."
)

from IPython.display import Audio, display

display(Audio(wavs[0], rate=sr))
```

```python
import soundfile as sf

sf.write("shorts_voice.wav", wavs[0], sr)

print("✅ Saved: shorts_voice.wav")
from google.colab import files

files.download("shorts_voice.wav")
```
